```
import pandas as pd

# Load the dataset
file_path = 'ultimate_student_productivity_dataset_5000(1).csv'
df = pd.read_csv(file_path)

# Display the first few rows
display(df.head())
```

```
df.info()
```

# Aggregate Statistics:

```
display(df.describe())
```

```
display(df['age'].describe())
```

# Filtering Data:

```
display(df[(df['age'] > 18) & (df['study_hours'] > 10)][['student_id', 'age', 'study_hours']])
```

```
display(df[(df['study_hours'] > 10)][['student_id', 'academic_level', 'study_hours']])
```

# Grouping Data:

```
display(df.groupby('gender')['productivity_score'].mean())
```

```
display(df.groupby('academic_level')['study_hours'].mean())
```

```
display(df.groupby('academic_level')['screen_time_hours'].mean())
```

```
import matplotlib.pyplot as plt

# Calculate counts for each academic level
academic_counts = df['academic_level'].value_counts()

# Create the pie chart
plt.figure(figsize=(8, 8))
plt.pie(academic_counts, labels=academic_counts.index, autopct='%1.1f%%', startangle=90)
plt.title('Distribution of Academic Levels')
plt.show()
```

```
import seaborn as sns
import matplotlib.pyplot as plt

# Create bar plot
plt.figure(figsize=(10, 6))
sns.barplot(x='age', y='gaming_hours', data=df)

# Add title and labels
plt.title('Age vs Gaming Hours')
plt.xlabel('Age')
plt.ylabel('Average Gaming Hours')
plt.show()
```
